```python

import spacy
from spacy import displacy
from IPython.display import HTML, display

#   Load the English language model
nlp = spacy.load("en_core_web_sm")

text = "I saw the man with a telescope."

doc = nlp(text)

# Print the dependency parse information for each token
for token in doc:
    print(f"{token.text:<12} {token.dep_:<12} {token.head.text}")

html = displacy.render(doc, style="dep", jupyter=False)

# Display the dependency parse visualization in Jupyter Notebook
display(HTML(html))

# Discuss with your partner:
# 1. The ROOT is “saw” because it is the main verb of the sentence.
# 2. spaCy most likely attaches “with a telescope” to “saw”, meaning the telescope was used for seeing.
# 3. Yes, the sentence is ambiguous
# 4. with a telescope” modifies the verb “saw”. 
# 5. “I saw the man who was carrying a telescope.”
     

import spacy

nlp = spacy.load("en_core_web_sm")

text = """
Microsoft announced a new AI research initiative
in Berlin on 15 June 2026.

The project involves collaboration with Google,
SAP and SRH University Heidelberg.

Satya Nadella will present the initiative
at a conference in Germany.
"""

doc = nlp(text)

for ent in doc.ents:
    print(
        ent.text,
        ent.label_,
        spacy.explain(ent.label_)
    )

#1. Microsoft
# Berlin
# 15 June 2026
# Google
# SAP
# SRH University Heidelberg
# Satya Nadella
# Germany
#2. Microsoft → ORG
# Berlin → GPE
# 15 June 2026 → DATE
# Google → ORG
# SAP → ORG
# SRH University Heidelberg → ORG
# Satya Nadella → PERSON
# Germany → GPE
#3. ORG: Companies, agencies, institutions, etc.
#4. I think none of the entities were misclassified. They all seem to be correctly labeled based on their context in the text.
     

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "AI is transforming healthcare",
    "Machine learning improves healthcare",
    "Football is a popular sport"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

similarity_matrix = cosine_similarity(tfidf_matrix)

print("\nCosine Similarity Matrix:")
print(similarity_matrix)

print("Doc1 vs Doc2:", round(similarity_matrix[0][1], 3))
print("Doc1 vs Doc3:", round(similarity_matrix[0][2], 3))
print("Doc2 vs Doc3:", round(similarity_matrix[1][2], 3))

# Questions:
# 1. Which pair has the highest similarity?
# 2. Is this result reasonable?
# 3. Why does TF-IDF treat some words as more important?
# 4. Does cosine understand meaning?
# 5. Would it know that "AI" and "machine learning" are related?

#Answers:
# 1. Doc1 vs Doc2 has the highest similarity.
# 2. Yes, this result is reasonable because both documents discuss healthcare and the impact of AI and machine learning on it, while Doc3 is about football, which is unrelated.
# 3.TF-IDF treats some words as more important based on their frequency in the document and across the corpus. Words that are common in a document but rare across the corpus are given higher weights, while common words across the corpus are down-weighted.
# 4.No, Cosine similarity measures the angle between vectors, not the semantic meaning of the text.
# 5.No, it would not know that "AI" and "machine learning" are related unless they appear in similar contexts in the documents, which would be reflected in their vector representations.
     
```
